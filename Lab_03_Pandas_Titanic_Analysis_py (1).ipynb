```python
# Problem 1: Dataset Inspection

# 1. Display first 10 rows
print("First 10 Rows:")
display(df.head(10))

# 2. Number of rows and columns
print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 3. Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# 4. Display data types
print("\nData Types:")
print(df.dtypes)

# 5. Descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())
```

```python
# Problem 2: Data Filtering

# 1. First class passengers
print("1. First Class Passengers:")
display(df[df["Pclass"] == 1])

# 2. Passengers younger than 18
print("\n2. Passengers Younger Than 18:")
display(df[df["Age"] < 18])

# 3. Male passengers who survived
print("\n3. Male Passengers Who Survived:")
display(df[
    (df["Sex"] == "male") &
    (df["Survived"] == 1)
])

# 4. Passengers who paid fare greater than 100
print("\n4. Passengers With Fare Greater Than 100:")
display(df[df["Fare"] > 100])
```

```python
# Problem 3: Sorting and Aggregation

# 1. Top 10 highest fares
print("1. Top 10 Passengers With Highest Fares:")
display(df.sort_values(
    by="Fare",
    ascending=False
).head(10))

# 2. Average passenger age
print("\n2. Average Passenger Age:")
print(df["Age"].mean())

# 3. Count male and female passengers
print("\n3. Number of Male and Female Passengers:")
print(df["Sex"].value_counts())

# 4. Average fare for each passenger class
print("\n4. Average Fare by Passenger Class:")
print(df.groupby("Pclass")["Fare"].mean())

# 5. Survival rate for each passenger class
print("\n5. Survival Rate by Passenger Class:")
print(df.groupby("Pclass")["Survived"].mean())
```

```python
# Final Laboratory Challenge
# Titanic Dataset Analysis

# 1. How many passengers are in the dataset?
total_passengers = len(df)

print("1. Total Number of Passengers:")
print(total_passengers)


# 2. What is the average passenger age?
average_age = df["Age"].mean()

print("\n2. Average Passenger Age:")
print(round(average_age, 2))


# 3. Which passenger class has the highest number of passengers?
passengers_by_class = df["Pclass"].value_counts()

print("\n3. Number of Passengers in Each Class:")
print(passengers_by_class)

highest_class = passengers_by_class.idxmax()

print("Passenger Class With Highest Number:")
print(highest_class)


# 4. What is the average fare paid by male and female passengers?
average_fare_gender = df.groupby("Sex")["Fare"].mean()

print("\n4. Average Fare by Gender:")
print(average_fare_gender)


# 5. Which gender has a higher survival rate?
survival_gender = df.groupby("Sex")["Survived"].mean()

print("\n5. Survival Rate by Gender:")
print(survival_gender)

higher_survival_gender = survival_gender.idxmax()

print("Gender With Higher Survival Rate:")
print(higher_survival_gender)


# 6. How does survival rate differ across passenger classes?
survival_class = df.groupby("Pclass")["Survived"].mean()

print("\n6. Survival Rate by Passenger Class:")
print(survival_class)
```

```python
# Convert survival rates into percentages

print("Survival Rate by Gender (%):")
print((survival_gender * 100).round(2))

print("\nSurvival Rate by Passenger Class (%):")
print((survival_class * 100).round(2))
```
