```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")

# Display first five rows
print(df.head())

# Dataset information
df.info()

# Display column names
print(df.columns)

# Display numerical columns
print("Numerical Attributes:")
print(df.select_dtypes(include="number").columns)

# Display categorical columns
print("Categorical Attributes:")
print(df.select_dtypes(include="object").columns)
```

```python
average_fare = df.groupby("Pclass")["Fare"].mean()

plt.plot(average_fare.index, average_fare.values)

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.show()
```

```python
plt.scatter(df["Age"], df["Fare"])

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
gender_count = df["Sex"].value_counts()

plt.bar(gender_count.index, gender_count.values)

plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.hist(df["Age"].dropna(), bins=10)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
survival_count = df["Survived"].value_counts()

plt.pie(
    survival_count,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)

plt.title("Passenger Survival Proportion")

plt.show()
```

```python
fig, ax = plt.subplots(1, 2)

# Histogram
ax[0].hist(df["Age"].dropna())
ax[0].set_title("Histogram of Passenger Ages")

# Bar chart
class_count = df["Pclass"].value_counts()

ax[1].bar(class_count.index, class_count.values)
ax[1].set_title("Number of Passengers by Class")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# Bar chart
gender_count = df["Sex"].value_counts()

ax[0].bar(gender_count.index, gender_count.values)
ax[0].set_title("Passengers by Gender")

# Histogram
ax[1].hist(df["Fare"], bins=10)
ax[1].set_title("Distribution of Passenger Fares")

# Pie chart
survival_count = df["Survived"].value_counts()

ax[2].pie(
    survival_count,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)
ax[2].set_title("Survival Proportion")

plt.tight_layout()
plt.show()
```
